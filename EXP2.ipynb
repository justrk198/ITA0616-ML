```python

import pandas as pd

# Step 1: Create training dataset
data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Strong'],
    'Water': ['Warm', 'Warm', 'Warm', 'Cool'],
    'Forecast': ['Same', 'Same', 'Change', 'Change'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

# Save dataset as CSV
df.to_csv('training_data.csv', index=False)

print("Training Dataset:")
print(df)

# Step 2: Initialize boundaries
attributes = list(df.columns[:-1])

S = ['0'] * len(attributes)
G = [['?'] * len(attributes)]

# Step 3: Candidate Elimination
for _, row in df.iterrows():

    example = list(row[:-1])
    target = row['EnjoySport']

    if target == 'Yes':

        # Remove inconsistent general hypotheses
        G = [g for g in G if all(
            g[i] == '?' or g[i] == example[i]
            for i in range(len(attributes))
        )]

        # Generalize specific hypothesis
        for i in range(len(attributes)):
            if S[i] == '0':
                S[i] = example[i]
            elif S[i] != example[i]:
                S[i] = '?'

    else:

        # Specialize general hypotheses
        new_G = []

        for g in G:
            if all(g[i] == '?' or g[i] == example[i]
                   for i in range(len(attributes))):

                for i in range(len(attributes)):
                    if S[i] != '?' and S[i] != '0':
                        if S[i] != example[i]:
                            new_h = g.copy()
                            new_h[i] = S[i]
                            new_G.append(new_h)
            else:
                new_G.append(g)

        G = new_G

    print("\nSpecific Boundary S:", S)
    print("General Boundary G:", G)

# Step 4: Final Result
print("\nFINAL SPECIFIC HYPOTHESIS:")
print(S)

print("\nFINAL GENERAL HYPOTHESES:")
for h in G:
    print(h)
```
